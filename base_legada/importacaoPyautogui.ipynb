```python
import pandas as pd
import pyautogui
import pyperclip
import time
import random
import os
import pyautogui
import time

from IPython.display import display, Javascript

usuarios = [        
    {'nome': '22597366880', 'senha': 'Amcham01!'},
    {'nome': '26967567837', 'senha': 'Amcham01!'},
    {'nome': '49043558818', 'senha': '100821Gf.'}   
]
coordenadas_por_usuario = {
    '26967567837': {
        'primeira_volta': (557, 607),
        'voltas_posteriores': (655, 617),
        'status': (337, 481)
    },
    '49043558818': {
        'primeira_volta': (557, 607-10),
        'voltas_posteriores': (655, 617-10),
        'status': (337, 481-10)
    },
    '48626200862': {
        'primeira_volta': (557, 607-14),
        'voltas_posteriores': (655, 618-14),
        'status': (337, 484-14)
    },
    '22597366880': {
        'primeira_volta': (576,782+63),
        'voltas_posteriores': (655, 605+63),
        'status': (337, 467+648)
    }
}


usuario_atual_idx = 0
falhas_consecutivas = 0
ordem_anterior = None

# 1. Carrega Excel original
df = pd.read_excel('importarNotas20250604.xlsm')
df['Status'] = df['Status'].astype(str)

# 2. Filtra linhas pendentes
pendentes = df[df['Status'].fillna('') != 'Enviado']

# 3. Arquivo de log
log_path = 'log_notas.txt'
start_time = time.time()
log_linhas = []

def salvar_log():
    with open(log_path, 'a', encoding='utf-8') as f:
        f.write('\n'.join(log_linhas) + '\n')
    log_linhas.clear()
    print("📝 Log salvo.")

# 4. Loop principal
primeira_volta = True
indice_atual = 0

def login_inicial(usuario):
    log_linhas.append("🚪 Iniciando login inicial...")

    # Clicar em botão de novo login
    pyautogui.moveTo(502, 579)
    pyautogui.click()
    time.sleep(5)

    # Clicar no campo de login
    pyautogui.moveTo(865, 587)
    pyautogui.click()
    time.sleep(1)
    
    # Digitar CPF
    pyautogui.write(usuario['nome'])
    pyautogui.press('enter')
    time.sleep(2)

    # Selecionar texto anterior (SHIFT+END, depois SHIFT+HOME)
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('del')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    pyautogui.press('backspace')
    time.sleep(1)    

    # Digitar senha
    pyautogui.write(usuario['senha'])
    pyautogui.press('enter')
    time.sleep(2)

    log_linhas.append(f"🔑 Login com usuário: {usuario['nome']}")

    # Acessar a página desejada
    pyautogui.moveTo(321, 82)
    pyautogui.click()
    time.sleep(1)
    pyautogui.write('https://www.nfp.fazenda.sp.gov.br/EntidadesFilantropicas/CadastroNotaEntidadeAviso.aspx')
    pyautogui.press('enter') 
    time.sleep(2)
    
    # clicar em prosseguir
    prosseguir()    

def prosseguir():
    # clicar em prosseguir
    pyautogui.moveTo(472,704) #prosseguir
    pyautogui.click()
    time.sleep(2)
    
    # clicar no combo de entidades e selecionar vocação
    pyautogui.moveTo(706,625+40) 
    pyautogui.click()
    time.sleep(1)
    pyautogui.moveTo(722,679+40) 
    pyautogui.click()
    time.sleep(1)
    
    # clicar em nova nota
    pyautogui.moveTo(552,755+40) 
    pyautogui.click()
    time.sleep(1)
    time.sleep(2)

# 5. Login inicial com primeiro usuário
login_inicial(usuarios[usuario_atual_idx])


while indice_atual < len(pendentes):
    idx = pendentes.index[indice_atual]
    chave = str(pendentes.loc[idx, 'Chave'])
    ordem = pendentes.loc[idx, 'Ordem']
    log_entry = f"🔄 Ordem {int(ordem)} - Chave: {chave}"
    

    print(log_entry)
    log_linhas.append(log_entry)
    usuario_logado = usuarios[usuario_atual_idx]['nome']
    coords = coordenadas_por_usuario[usuario_logado]

    # Cola a chave no campo
    if primeira_volta:
        #pyautogui.moveTo(557, 607) # Log Felipe
        #pyautogui.moveTo(557, 594) # Log gaby
        #pyautogui.moveTo(557, 594) # Log Muma
        #pyautogui.moveTo(557, 607) # Log manu
        pyautogui.moveTo(*coords['primeira_volta'])
        primeira_volta = False
    else:
        #pyautogui.moveTo(655, 618) # Log felipe
        #pyautogui.moveTo(655, 605) # Log gaby
        #pyautogui.moveTo(655, 605) # Log Muma
        #pyautogui.moveTo(655, 618) # Log manu
        pyautogui.moveTo(*coords['voltas_posteriores'])

    pyautogui.click()
    pyperclip.copy(chave)
    pyautogui.hotkey('ctrl', 'v')
    pyautogui.press('enter')
    #time.sleep(random.uniform(1, 4))
    time.sleep(1) #aumenta tempo da volta

    # Copia status
    #pyautogui.moveTo(337, 484) # Log felipe
    #pyautogui.moveTo(337, 467) # Log gaby
    #pyautogui.moveTo(337, 467) # Log Muma
    #pyautogui.moveTo(337, 484) # Log manu
    pyautogui.moveTo(*coords['status'])
    pyautogui.click(); pyautogui.click()
    pyautogui.hotkey('ctrl', 'c')
    time.sleep(0.5)
    status = pyperclip.paste().strip()
    #status = "não"
    log_linhas.append(f"📋 Status: '{status}'")

    status_lower = status.lower()
    ordem_atual = pendentes.loc[idx, 'Ordem']

    # Lógica de falha "Não"
    if "não" in status_lower:
        
        if ordem_anterior is None or ordem_atual == ordem_anterior + 1:
            falhas_consecutivas += 1
            print("Adiciona falha")
        else:
            falhas_consecutivas = 1  # Reset se a ordem não for sequencial
            print("Reset falha")
    
        ordem_anterior = ordem_atual
        print(falhas_consecutivas)
        log_linhas.append(f"⚠️ Falha consecutiva detectada: {falhas_consecutivas} com ordem {ordem_atual}")
        print(falhas_consecutivas)
        if falhas_consecutivas >= 6:
            print('Limite de falhas alcançado.Trocando usuário...')
            log_linhas.append("🔁 Limite de falhas alcançado. Trocando usuário...")
    
            # Troca usuário
            usuario_atual_idx = (usuario_atual_idx + 1) % len(usuarios)
            usuario_novo = usuarios[usuario_atual_idx]
    
            # Encerra sessão
            pyautogui.moveTo(196, 310)
            pyautogui.click()
            time.sleep(30)
    
            # Reutiliza o login inicial
            login_inicial(usuario_novo)    
    
            log_linhas.append(f"🔑 Novo usuário: {usuario_novo['nome']}")
            print(f"🔑 Novo usuário: {usuario_novo['nome']}")
            # Volta 20 posições
            indice_atual = max(indice_atual - 20, 0)
            falhas_consecutivas = 0
            primeira_volta = True

            #clicar em endereço e digitar https://www.nfp.fazenda.sp.gov.br/EntidadesFilantropicas/CadastroNotaEntidadeAviso.aspx
            pyautogui.moveTo(321, 82)
            pyautogui.click()
            time.sleep(1)
            pyautogui.write('https://www.nfp.fazenda.sp.gov.br/EntidadesFilantropicas/CadastroNotaEntidadeAviso.aspx')
            pyautogui.press('enter') 
            time.sleep(2)

            # clicar em prosseguir
            prosseguir()

    if "sucesso" in status_lower:
        df.at[idx, 'Status'] = 'Enviado'
        indice_atual += 1
        falhas_consecutivas = 0
    elif "pedido" in status_lower or "já cadastrada" in status_lower:
        df.at[idx, 'Status'] = 'Duplicada'
        indice_atual += 1
        falhas_consecutivas = 0
    elif status_lower == "":
        log_linhas.append("⚠️ Status vazio. Reiniciando e voltando 5 posições.")
        df.at[idx, 'Status'] = 'inicio'

        pyautogui.moveTo(574,17) 
        pyautogui.click()
        
        #clicar em endereço e digitar https://www.nfp.fazenda.sp.gov.br/EntidadesFilantropicas/CadastroNotaEntidadeAviso.aspx
        pyautogui.moveTo(321, 82)
        pyautogui.click()
        time.sleep(1)
        pyautogui.write('https://www.nfp.fazenda.sp.gov.br/EntidadesFilantropicas/CadastroNotaEntidadeAviso.aspx')
        pyautogui.press('enter') 
        time.sleep(2)
        
        # clicar em prosseguir
        prosseguir()

        indice_atual = max(indice_atual - 5, 0)
        primeira_volta = True
    else:
        erro = f"Erro: {status[:30]}"
        log_linhas.append(f"❌ {erro}")
        df.at[idx, 'Status'] = erro
        indice_atual += 1
        #falhas_consecutivas = 0

    time.sleep(1)
    pyautogui.moveTo(1914,1020)
    pyautogui.click()
    pyautogui.moveTo(1887,653)
    pyautogui.click()
   

    # Salva log a cada 5 minutos
    if time.time() - start_time >= 300:  # 5 minutos = 300 segundos
        salvar_log()
        start_time = time.time()

# Salvamento final do log ao fim do processo
if log_linhas:
    salvar_log()

print("✅ Processo concluído e log salvo.")
```
